## Student Performance

#### Q1. Finding  Mean, Mdian, Modd for (Maths Marks)

```python
import numpy as np
import pandas as pd
import random
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
SP_data = pd.read_csv(r"../student.csv")
SP_data.head(3)
```

```python
total_mean=np.mean(SP_data[["Math"]])
print("Mean of Math Marks",total_mean)
```

```python
SP_data["Study_Hours"].mean()
```

```python
np.median(SP_data['Study_Hours'])
```

```python
total_medi =np.median(SP_data[["Math"]])
print("Median of Math Marks",total_medi)
```

```python
SP_data["Math"].mode()
```

```python
SP_data["Math"].mode()[0]
```

```python
#Frequencies

SP_data["Math"].value_counts()
```

```python
#Range
min_r = SP_data['Math'].min()
max_r = SP_data['Math'].max()
range =max_r- min_r
Std = SP_data['Math'].std()
print("Min Range:-",min_r)
print("Max Range:-",max_r)
print("Standard Deviation:-",Std)
print('Range:-',range)
```

```python
#Mean Absolute Devation
data =SP_data['Math']
mad =np.sum(np.abs(data-np.mean(data)))/len(data)
print('Mean Absolute Devation:-',mad)
```

```python
#Quartiles & IOR
q1=np.percentile(SP_data['Math'],25)
q2=np.percentile(SP_data['Math'],50)
q3=np.percentile(SP_data['Math'],75)
iqr = q3-q1
print(q1,q2,q3)
print('IQR:-',iqr)
```
